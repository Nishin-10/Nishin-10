```python
from matplotlib import pyplot as plt
import cv2
import numpy as np
```

```python
filename = '/content/drive/MyDrive/v_data/istockphoto-187536877-612x612.jpg'
img = cv2.imread(filename)
plt.imshow(img)
```

```python
#interpolation
(height, width) = img.shape[:2]
res = cv2.resize(img, None, fx=2, fy=2, interpolation = cv2.INTER_CUBIC)
plt.imshow(res)
```

```python
#Rotation
rows,cols = img.shape[:2]
M = cv2.getRotationMatrix2D((cols/2, rows/2), 90, 1)
rotated_img = cv2.warpAffine(img, M, (cols, rows))
plt.imshow(rotated_img)
```

```python
# Shifting

M =np.float32([[1,0,100],[0,1,50]])
rows, cols = img.shape[:2]
res =cv2.warpAffine (img, M, (cols, rows))
shifted_img = cv2.warpAffine(img, M, (cols, rows))
plt.imshow(shifted_img)
```

```python
# Line detection
res = cv2.Canny(img, 100, 200)
plt.imshow(res)
```

```python
#blur kernel
kernel = np.ones ((5,5),np.float32)/25
res =cv2.filter2D (img, -1, kernel)
plt.imshow(res)
```

```python
#blur median
res = cv2.medianBlur(img, 5)
plt.imshow(res)
```

```python
# blur bilateral
res = cv2.bilateralFilter(img, 9, 75, 75)
plt.imshow (res)
```

```python
# Sharpen the Image
kernel = np.array (([-1,-1,-1],[-1,9,3],[-1,-1,-1]))
res = cv2.filter2D(img, -1, kernel)
plt.imshow(res)
```

```python
#convert into BW
res = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(res)
```
